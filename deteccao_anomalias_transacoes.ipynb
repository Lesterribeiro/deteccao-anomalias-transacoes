# Detecção de Anomalias em Transações Financeiras

Este notebook realiza a análise, pré-processamento, aplicação de técnicas de balanceamento de dados (`SMOTE`) e treinamento do algoritmo **Random Forest Classifier** para identificação de fraudes/anomalias em transações financeiras conforme o projeto da Jornada Tech.

---

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# 1. Geração / Carregamento dos dados desbalanceados
X, y = make_classification(
    n_samples=284807, 
    n_features=30, 
    n_informative=20, 
    n_redundant=10, 
    weights=[0.9983, 0.0017], 
    random_state=42
)

# Separação em treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Dimensão do Treino: {X_train.shape}")
print(f"Dimensão do Teste: {X_test.shape}")
```

---
## 2. Avaliação e Técnicas de Balanceamento
Aplicação do SMOTE no conjunto de treino para balanceamento das classes.

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X_train, y_train)

print(f"Resampled shape: {X_res.shape}")
```

---
## 3. Modelos Avançados e Explicabilidade
Treinamento do modelo `RandomForestClassifier` exatamente como estruturado no projeto.

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=50,
    max_depth=10,
    class_weight="balanced",
    n_jobs=-1,
    random_state=42
)

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

print(classification_report(y_test, y_pred_rf))
```
